B5. Airflow Component Flow Diagram
Flow:

DAG File → Scheduler → Executor / Worker → Metadata Database → Webserver (UI)

DAG File: Contains the workflow and tasks.

Scheduler: Checks the DAG and decides when to run the tasks.

Executor / Worker: Runs the tasks.

Metadata Database: Stores information about the DAGs and tasks.

Webserver (UI): Shows the DAGs, task status, logs, and results.
