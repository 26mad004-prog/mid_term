```python
import pandas as pd
import numpy as np
```

```python
emp = pd.DataFrame({
    'name':["Ania", 'Bharat', 'Chetan', 'Divya', "Esha", 'Farhan','Gaurav' ],
    'dept':['Eng', 'Eng', 'Ana', 'Ana','Ana','DP','DP'],
    'Salary':[90000, 60000, 75000, 55000, 62000, 82000,50000]
})
```

```python

avg = (
    emp.groupby('dept', as_index=False)['Salary']
       .mean()
       .rename(columns={'Salary': 'avg_salary'})
)
b = emp.merge(avg, on='dept', how='left')

print(b)
```

```python
cte1 = (
    emp.groupby('dept', as_index=False)['Salary']
       .mean()
       .rename(columns={'Salary': 'dept_avg'})
)

cte2 = emp.merge(cte1, on='dept', how='left')

result = cte2[cte2['Salary'] > cte2['dept_avg']].copy()

result['pct_above_avg'] = (
    (result['Salary'] - result['dept_avg']) / result['dept_avg'] * 100
)

print(result)
```

```python
cte1 = (
    emp.groupby('dept', as_index=False)['Salary']
       .mean()
       .rename(columns={'Salary': 'dept_avg'})
)

cte2 = emp.merge(cte1, on='dept')

result = (
    cte2[cte2['Salary'] > cte2['dept_avg']]
    .assign(
        pct_above_avg=lambda x:
            (x['Salary'] - x['dept_avg']) / x['dept_avg'] * 100
    )
    .sort_values('pct_above_avg', ascending=False)
)

print(result)
```
